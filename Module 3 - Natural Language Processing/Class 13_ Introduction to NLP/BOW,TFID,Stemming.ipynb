##Token Normalizing / Data Clearning

Stemming

```python
from nltk.stem import PorterStemmer
import re
```

```python
stemmer = PorterStemmer()
sentences = nltk.sent_tokenize(corpus)
print(sentences)
```

```python
stem_sentences = []
for i in range(len(sentences)):
  text = re.sub('[^a-zA-Z0-9]',' ', sentences[i])
  text = text.lower()
  words = nltk.word_tokenize(text)
  words = [stemmer.stem(word) for word in words]
  stem_sentences.append(' '.join(words))
stem_sentences
```

Lemmatization


```python
from nltk.stem import WordNetLemmatizer
import re
nltk.download('wordnet')
```

```python
lematizer = WordNetLemmatizer()

sentences_lem = []
for i in range(len(sentences)):
  text = re.sub('[^a-zA-Z0-9]',' ', sentences[i])
  text = text.lower()
  words = nltk.word_tokenize(text)
  words = [lematizer.lemmatize(word) for word in words]
  sentences_lem.append(' '.join(words))
sentences_lem
```

Sentiment analysis using Bag of Words tokener & Logistic Regression

```python
# Import necessary libraries
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

```

```python
# Step 1: Load plain text IMDb movie review dataset
# Assume the dataset has two columns: 'review' and 'sentiment' (0 = Negative, 1 = Positive)
# You can load this from a CSV, or directly from a Kaggle download
# Here, we use a small custom dataset for demonstration
```

```python
data = {
    'review': [
        "The movie was fantastic, with brilliant performances!",
        "I did not like the movie at all. It was boring and too long.",
        "The plot was good, but the pacing was very slow.",
        "What a waste of time! Absolutely terrible.",
        "An amazing movie with a touching story.",
        "Mediocre at best, nothing special in this film.",
        "The acting was subpar, and the plot was predictable."
    ],
    'sentiment': [1,0,0,0,1,0,0] # 1 = Positive, 0 = Negative
}
```

```python
# Converting data into DataFrame
df = pd.DataFrame(data)
# Step 2: Split the dataset into training and test sets
x_train, x_test, y_train, y_test = train_test_split(df['review'], df['sentiment'], test_size=0.2, random_state=42)
# Step 3: Feature extraction - N-grams with BoW or TF-IDF
# BoW with unigram and bigram

vectorizer = CountVectorizer(ngram_range=(1,2), max_features=10)
# For TF-IDF with unigrams and bigrams, you can switch to this:
# vectorizer = TfidfVectorizer(ngram_range=(1, 2), max_features=1000)

x_train_vectorized = vectorizer.fit_transform(x_train)
x_test_vectorized = vectorizer.transform(x_test)

# Step 4: Train Logistic Regression model
model = LogisticRegression(max_iter=1000)
model.fit(x_train_vectorized, y_train)

# Step 5: Make predictions and evaluate the model
y_pred = model.predict(x_test_vectorized)
accuracy = accuracy_score(y_test, y_pred)

print(f"Model Accuracy: {accuracy * 100:.2f}%")
```

```python
# Step 6: Predict sentiments for new reviews
custom_reviews = [
    "I loved the movie. It had a gripping storyline and fantastic acting.",
    "It was one of the worst films I've ever seen. Absolutely terrible.",
    "An okay movie, not great but not bad either.",
    "The movie had some strong performances but was overall too long."
]
```

```python
# Vectorizing custom reviews
custom_reviews_vectorized = vectorizer.transform(custom_reviews)

# Making predictions
custom_predictions = model.predict(custom_reviews_vectorized)

# Mapping predictions to sentiment labels
sentiments = ['Negative', 'Positive']
for review, prediction in zip(custom_reviews, custom_predictions):
    print(f"Review: {review}\nSentiment: {sentiments[prediction]}\n")
```
